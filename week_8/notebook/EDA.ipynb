```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"../data/cleaned/timestap_pjme.csv")
df['Datetime'] = df['Datetime'].astype('datetime64[ms]')
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data=df, x="hour", y="PJME_MW",linewidth=0.5)
plt.title("Time Series Visualization")
plt.xlabel("Hours")
plt.ylabel("Mega Watt")
plt.grid(alpha=0.3)
plt.xticks(rotation=0) 
plt.tight_layout()
plt.show()
```

```python
df['week_of_year'] = df['Datetime'].dt.isocalendar().week
```

```python
#weekly demand
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data=df, x="week_of_year", y="PJME_MW",linewidth=0.5)
plt.title("weekly for year  Visualization")
plt.xlabel("weeek")
plt.ylabel("Mega Watt")
plt.grid(alpha=0.3)
plt.xticks(rotation=0) 
plt.tight_layout()
plt.show()
```

```python
#daily demand
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data=df, x="day_name", y="PJME_MW",linewidth=0.5)
plt.title("Day Series Visualization")
plt.xlabel("day")
plt.ylabel("Mega Watt")
plt.grid(alpha=0.3)
plt.xticks(rotation=0) 
plt.tight_layout()
plt.show()
```

```python
#montly
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data=df, x="month", y="PJME_MW",linewidth=0.5)
plt.title("Monthly Series Visualization")
plt.xlabel("month")
plt.ylabel("Mega Watt")
plt.grid(alpha=0.3)
plt.xticks(rotation=0) 
plt.tight_layout()
plt.show()
```

```python
df['day_of_month'] = df['Datetime'].dt.day
```

```python
#heat map
```

```python
pivot_df = df.pivot_table(index="day_of_month", columns="hour", values="PJME_MW", aggfunc="mean")
```

```python
plt.figure(figsize=(15, 8)) 

sns.heatmap(pivot_df,cmap="YlGnBu",annot=True,fmt=".0f", annot_kws={"size":7},linewidths=0.5)

plt.title("Heatmap by Day and Hour")
plt.xlabel("Hour of the Day (0-23)")
plt.ylabel("Day of the Week")
plt.xticks(rotation=0) 
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
