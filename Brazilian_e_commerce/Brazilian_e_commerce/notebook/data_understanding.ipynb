```python
#merging the data frames 

#aggregating the payment

df_pay_agg = df_order_pay.groupby("order_id").agg(
    total_payment_value=("payment_value","sum"),
    max_installments=("payment_installments","max"),
    primary_payment_type =("payment_type",lambda x:x.mode()[0] if not x.empty else "unknown")
).reset_index()

df_review_agg = df_order_review.groupby("order_id").agg(
    avg_review_score =("review_score","mean"),
    review_count=("review_id","count")
).reset_index()

df_product_translated = df_product_data.merge(df_category,on ="product_category_name",
                                              how="left")

df_master_raw = (
    df_order_item
    .merge(df_order_dataset, on='order_id', how='inner')
    .merge(df_customer, on='customer_id', how='left')
    .merge(df_product_translated, on='product_id', how='left')
    .merge(df_seller, on='seller_id', how='left')
    .merge(df_pay_agg, on='order_id', how='left')
    .merge(df_review_agg, on='order_id', how='left')
)

df_master_raw
```

```python
#import matplotlib and seaborn

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Missing Value Heatmap
plt.figure(figsize=(12, 12))
sns.heatmap(df_master_raw.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap in Master Raw Dataset')
plt.tight_layout()
plt.show()

# 2. Data Type Summary
plt.figure(figsize=(6, 6))
df_master_raw.dtypes.value_counts().plot(kind='bar', color='teal')
plt.title('Column Data Types Count')
plt.ylabel('Count')
plt.xlabel('Data Type')
plt.tight_layout()
plt.show()
```

```python
table_sizes = pd.DataFrame({
    'Dataset': [
        'Customers','Orders','Order Items',
        'Products','Sellers','Payments',
        'Reviews','Geolocation','Category Translation'
    ],
    'Rows': [
        df_customer.shape[0],df_order_dataset.shape[0],df_order_item.shape[0],
        df_product_data.shape[0],df_seller.shape[0],df_order_pay.shape[0],
        df_order_review.shape[0],df_geolocation.shape[0],df_product_translated.shape[0]  
    ]})

plt.figure(figsize=(15,6))

ax = sns.barplot(data=table_sizes,x="Dataset",y="Rows")

for container in ax.containers:
    ax.bar_label(container, fmt='%d')

plt.title("Table Size Comparison", fontsize=16, fontweight='bold')
plt.xlabel("Datasets")
plt.ylabel("Number of Rows")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
