```python
#eda analysis
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv(r"../data/selected/selcetd_train_x.csv")

df_y = pd.read_csv(r"../data/selected/train_y.csv")
```

```python
merger_df = pd.concat([df,df_y],axis=1)

corr_matrix = merger_df.corr()
```

```python
plt.figure(figsize=(26,22))

sns.heatmap(corr_matrix,annot=True,cmap="crest",linewidths=0.5)

plt.title("correlation matrix for the train data set")
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("../static/images/correlation.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
#scatter plot
plt.figure(figsize =(5,5))
sns.scatterplot(data=merger_df,x="TotalSf",y="SalePrice")
plt.title("linear relation")
plt.xlabel("totalsf")
plt.ylabel("sales price")
plt.tight_layout()
plt.savefig("../static/images/totalsf-linear.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize =(5,5))
sns.scatterplot(data=merger_df,x="TotalLivArea",y="SalePrice")
plt.title("linear relation")
plt.xlabel("totallivarea")
plt.ylabel("sales price")
plt.tight_layout()
plt.savefig("../static/images/totallivarea.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.histplot(data=merger_df,x='LotFrontage',kde=True,bins=30)
plt.title(f"disttibution of lot front area")
plt.xlabel("lot arae")
plt.ylabel("Frequency")
plt.savefig("../static/images/lotfrontage.png", dpi=300, bbox_inches="tight")
plt.tight_layout()
plt.show()
```
