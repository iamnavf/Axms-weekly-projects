```python
import pandas as pd
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, GlobalAveragePooling1D, Dense, Dropout, SimpleRNN, LSTM, GRU, Bidirectional
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
x_train = np.load(r"../data/cleaned/x_train.npy")
x_test = np.load(r"../data/cleaned/x_test.npy")
x_val = np.load(r"../data/cleaned/x_val.npy")
y_train = np.load(r"../data/cleaned/y_train.npy")
y_test = np.load(r"../data/cleaned/y_test.npy")
y_val = np.load(r"../data/cleaned/y_val.npy")
```

```python
x_train.shape,x_test.shape,x_val.shape
```

```python
y_train.shape,y_test.shape,y_val.shape
```

```python
x_train,x_test,x_val
```

```python
y_train,y_test,y_val
```

```python
VOCAB_SIZE = 20000
MAX_LEN = 200
EMBEDDING_DIM = 100
```

```python
#basic ann model
```

```python
ann_model = Sequential([
    Input(shape=(MAX_LEN,)),
    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=EMBEDDING_DIM
    ),
    GlobalAveragePooling1D(),
    Dense(64, activation="relu"),
    Dropout(0.3),
    Dense(1, activation="sigmoid")
])
```

```python
ann_model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
ann_model.fit(x_train,y_train, validation_data=(x_val,y_val),epochs=10,batch_size=32)
```

```python
y_pred = ann_model.predict(x_val)
```

```python
y_pred = (y_pred >= 0.5).astype(int).ravel()
```

```python
accuracy_ann = accuracy_score(y_val, y_pred)
precision_ann = precision_score(y_val, y_pred)
recall_ann = recall_score(y_val, y_pred)
f1_ann = f1_score(y_val, y_pred)
auc_score = roc_auc_score(y_val, y_pred)

print("Accuracy :", accuracy_ann)
print("Precision:", precision_ann)
print("Recall   :", recall_ann)
print("F1 Score :", f1_ann)
print("auc_score :",auc_score)
```

```python
ann_model.save(r"../model/base_ann.keras")
```

```python
baseline_performance = {"MODEL":[],
                        "ACCURACY":[],
                        "PRECISION":[],
                        "RECALL":[],
                        "F1 SCORE":[],
                        "AUC":[]}
```

```python
def performance(name,acc,precision,recall,f1,auc):
    baseline_performance["MODEL"].append(name),
    baseline_performance["ACCURACY"].append(acc),
    baseline_performance["PRECISION"].append(precision),
    baseline_performance["RECALL"].append(recall),
    baseline_performance["F1 SCORE"].append(f1),
    baseline_performance["AUC"].append(auc)
```

```python
performance("ANN",0.87,0.85,0.91,0.88,0.87)
```
