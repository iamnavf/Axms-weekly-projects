```python
import pandas as pd
import numpy as np
import joblib
import time


from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,average_precision_score
```

```python
top_15_indices = joblib.load(r"../models/top_15_indices.pkl")
```

```python
x_train = np.load(r"../data/processed/x_train_fea.npy")
x_test = np.load(r"../data/processed/x_test_fea.npy")
y_train = np.load(r"../data/processed/y_train_fea.npy")
y_test = np.load(r"../data/processed/y_test_fea.npy")
```

```python
x_train = x_train[:, top_15_indices]
x_test = x_test[:, top_15_indices]

print("Selected:", x_train.shape,x_test.shape)
```

```python
ann_model = Sequential([
    Dense(64, activation='relu', input_shape=(15,)),
    Dropout(0.3),

    Dense(32, activation='relu'),
    Dropout(0.3),

    Dense(1, activation='sigmoid')
])

ann_model.summary()
```

```python
ann_model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
start_time = time.time()
history = ann_model.fit(x_train,y_train,validation_split=0.2,epochs=10,batch_size=32)
training_time = time.time() - start_time
print("Training time:", training_time, "seconds")
```

```python
y_prob_ann = ann_model.predict(x_test).ravel()

y_pred_ann = (y_prob_ann >= 0.5).astype(int)
```

```python
print("accuracy :" ,accuracy_score(y_test, y_pred_ann))
print("precison: ", precision_score(y_test, y_pred_ann))
print("recall: ",recall_score(y_test, y_pred_ann))
print("f1 score :",f1_score(y_test, y_pred_ann))
print("roc_auc :", roc_auc_score(y_test, y_prob_ann))
print("pr_auc: ",average_precision_score(y_test, y_prob_ann))
```
