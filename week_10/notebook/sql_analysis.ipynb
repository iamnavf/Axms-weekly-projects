```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect('customer_intelligence.db')
```

```python
df = pd.read_csv(r"../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.to_sql('customers',conn,if_exists='replace',index=False)
```

```python
#total customer
```

```python
query = """SELECT COUNT(*) AS total_customers FROM customers;"""
result = pd.read_sql(query, conn)
result
```

```python
#total churened customer
```

```python
query = """SELECT COUNT(Churn) AS total_customers FROM customers where Churn = "Yes";"""
result = pd.read_sql(query, conn)
result
```

```python
#churn rate
```

```python
query = """SELECT (SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0/ COUNT(*)) AS churn_rate FROM customers;"""
result = pd.read_sql(query, conn)
result
```

```python
#chrun by contract
```

```python
query = """SELECT Contract,COUNT(*) AS churned_customers
FROM customers
WHERE Churn = 'Yes'
GROUP BY Contract
ORDER BY churned_customers DESC;"""
result = pd.read_sql(query, conn)
result
```

```python
#payment method
```

```python
query =""" SELECT  PaymentMethod,COUNT(*) AS CHURENED_CUSTOMER
FROM customers
WHERE Churn = "Yes"
GROUP BY PaymentMethod
ORDER BY CHURENED_CUSTOMER DESC;"""
result= pd.read_sql(query, conn)
result
```

```python
#churn by internet service
```

```python
query =""" SELECT  InternetService,COUNT(*) AS CHURENED_CUSTOMER
FROM customers
WHERE Churn = "Yes"
GROUP BY InternetService
ORDER BY CHURENED_CUSTOMER DESC;"""
result= pd.read_sql(query, conn)
result
```

```python
#average mothly charges
```

```python
query = """ SELECT AVG(MonthlyCharges) AS AVERAGE_CHARGES
FROM customers;"""
result = pd.read_sql(query,conn)
result
```

```python
#total charges
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
query = """ SELECT AVG(TotalCharges) AS AVERAGE_CHARGES
FROM customers;"""
result = pd.read_sql(query,conn)
result
```

```python
#high values customers
```

```python
query = """SELECT customerID,Contract,tenure,MonthlyCharges,TotalCharges
FROM customers
ORDER BY TotalCharges DESC
LIMIT 10;"""
result = pd.read_sql(query, conn)
result
```

```python
#high risk customers
```

```python
query = """SELECT customerID,Contract,tenure,MonthlyCharges,TotalCharges,Churn
FROM customers
WHERE Contract = 'Month-to-month'
  AND MonthlyCharges >= 70
  AND Churn = "No"
ORDER BY MonthlyCharges DESC
LIMIT 10;"""
result = pd.read_sql(query, conn)
result
```

```python
#customer with long tenure
```

```python
query = """SELECT customerID,Contract,tenure,MonthlyCharges,TotalCharges,Churn
FROM customers
ORDER BY tenure DESC
LIMIT 10;"""
result = pd.read_sql(query, conn)
result
```

```python
#customer with high monlthy cahrges
```

```python
query = """ SELECT customerID,Contract,tenure,MonthlyCharges,TotalCharges,Churn
FROM customers
ORDER BY MonthlyCharges DESC
LIMIT 10;"""
result = pd.read_sql(query, conn)
result
```

```python
#customer segment based on service
```

```python
query = """SELECT InternetService,COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS churn_rate
FROM customers
GROUP BY InternetService
ORDER BY total_customers DESC;"""
result = pd.read_sql(query, conn)
result
```

```python
#revenue analysis
```

```python
query =""" SELECT ROUND(SUM(MonthlyCharges),2) AS TOTAL_MONTHLY_CHARGES,
ROUND(AVG(MonthlyCharges),2) AS AVERAGE_MONTHLY_CHARGES
FROM customers;"""
result = pd.read_sql(query, conn)
result
```
