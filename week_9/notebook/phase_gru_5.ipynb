```python
import pandas as pd
import numpy as np

from tensorflow.keras import optimizers
from tensorflow.keras.callbacks import EarlyStopping,ReduceLROnPlateau
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, GlobalAveragePooling1D, Dense, Dropout, BatchNormalization,GRU
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
x_train = np.load(r"../data/cleaned/x_train.npy")
x_test = np.load(r"../data/cleaned/x_test.npy")
x_val = np.load(r"../data/cleaned/x_val.npy")
y_train = np.load(r"../data/cleaned/y_train.npy")
y_test = np.load(r"../data/cleaned/y_test.npy")
y_val = np.load(r"../data/cleaned/y_val.npy")
```

```python
x_train.shape,x_test.shape,x_val.shape
```

```python
VOCAB_SIZE = 20000
MAX_LEN = 200
```

```python
#emdeding
```

```python
gru_model = Sequential([
    Input(shape=(MAX_LEN,)),
    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=100
    ),
    GRU(64),
    Dense(64, activation="relu"),
    Dense(1, activation="sigmoid")
])
```

```python
gru_model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
gru_model.fit(x_train,y_train, validation_data=(x_val,y_val),epochs=10,batch_size=32)
```

```python
y_pred = gru_model.predict(x_val)
y_pred = (y_pred >= 0.5).astype(int).ravel()
```

```python
accuracy_gru = accuracy_score(y_val, y_pred)
precision_gru = precision_score(y_val, y_pred)
recall_gru = recall_score(y_val, y_pred)
f1_gru = f1_score(y_val, y_pred)
auc_score_gru = roc_auc_score(y_val, y_pred)

print("Accuracy :", accuracy_gru)
print("Precision:", precision_gru)
print("Recall   :", recall_gru)
print("F1 Score :", f1_gru)
print("auc_score :",auc_score_gru)
```

```python
gru_model.save(r"/content/drive/MyDrive/cleaned/gru_em100.keras")
```

```python
#sequence tuning
```

```python
x_trainseq = np.load(r"/content/drive/MyDrive/seq/x_train_seq.npy",allow_pickle=True)
x_testseq = np.load(r"/content/drive/MyDrive/seq/x_test_seq.npy",allow_pickle=True)
x_valseq = np.load(r"/content/drive/MyDrive/seq/x_val_seq.npy",allow_pickle=True)
```

```python
def create_sequences(length):
    x_train_pad = pad_sequences(
        x_trainseq,
        maxlen=length,
        padding="post",
        truncating="post"
    )

    x_val_pad = pad_sequences(
        x_valseq,
        maxlen=length,
        padding="post",
        truncating="post"
    )

    x_test_pad = pad_sequences(
        x_testseq,
        maxlen=length,
        padding="post",
        truncating="post"
    )
    return x_train_pad, x_val_pad, x_test_pad
```

```python
x_train_100, x_val_100, x_test_100 = create_sequences(100)
```

```python
gru_seq_100 = Sequential([
    Input(shape=(MAX_LEN,)),
    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=128
    ),
    GRU(128),
    Dense(64, activation="relu"),
    Dense(1, activation="sigmoid")
])
```

```python
gru_seq_100.compile(optimizer=optimizers.Adam(),loss="binary_crossentropy",metrics=["accuracy"])
history = gru_seq_100.fit(x_train_100,y_train, validation_data=(x_val_100,y_val),epochs=10,batch_size=32)
```

```python
y_pred = gru_seq_100.predict(x_val_100)
y_pred = (y_pred >= 0.5).astype(int).ravel()
```

```python
accuracy_gru = accuracy_score(y_val, y_pred)
precision_gru = precision_score(y_val, y_pred)
recall_gru = recall_score(y_val, y_pred)
f1_gru = f1_score(y_val, y_pred)
auc_score_gru = roc_auc_score(y_val, y_pred)

print("Accuracy :", accuracy_gru)
print("Precision:", precision_gru)
print("Recall   :", recall_gru)
print("F1 Score :", f1_gru)
print("auc_score :",auc_score_gru)
```
