```python
import pandas as pd
import numpy as np

from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import optimizers
from tensorflow.keras.callbacks import EarlyStopping,ReduceLROnPlateau
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, GlobalAveragePooling1D, Dense, Dropout, BatchNormalization,SimpleRNN
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
x_train = np.load(r"../data/cleaned/x_train.npy")
x_test = np.load(r"../data/cleaned/x_test.npy")
x_val = np.load(r"../data/cleaned/x_val.npy")
y_train = np.load(r"../data/cleaned/y_train.npy")
y_test = np.load(r"../data/cleaned/y_test.npy")
y_val = np.load(r"../data/cleaned/y_val.npy")
```

```python
VOCAB_SIZE = 20000
MAX_LEN = 200
```

```python
x_train.shape,x_test.shape,x_val.shape
```

```python
y_train.shape,y_test.shape,y_val.shape
```

```python
#embedded dimension
```

```python
rnn_model_em100 = Sequential([
    Input(shape=(MAX_LEN,)),
    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=100
    ),
    SimpleRNN(64),
    Dense(64, activation="relu"),
    Dense(1, activation="sigmoid")
])
```

```python
rnn_model_em100.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
history = rnn_model_em100.fit(x_train,y_train, validation_data=(x_val,y_val),epochs=10,batch_size=32)
```

```python
y_pred = rnn_model_em100.predict(x_val)
```

```python
y_pred = (y_pred >= 0.5).astype(int).ravel()
accuracy_rnn = accuracy_score(y_val, y_pred)
precision_rnn = precision_score(y_val, y_pred)
recall_rnn = recall_score(y_val, y_pred)
f1_rnn = f1_score(y_val, y_pred)
auc_score_rnn = roc_auc_score(y_val, y_pred)

print("Accuracy :", accuracy_rnn)
print("Precision:", precision_rnn)
print("Recall   :", recall_rnn)
print("F1 Score :", f1_rnn)
print("auc_score :",auc_score_rnn)
```

```python
rnn_model_em100.save(r"../models/rnn_model_em100.keras")

history_df =pd.DataFrame(history.history)

history_df.to_csv(r'../log/rnn_em100.csv',index=False)
```

```python
x_trainseq = np.load(r"../data/seq/x_train_seq.npy", allow_pickle=True)
x_testseq = np.load(r"../data/seq/x_test_seq.npy", allow_pickle=True)
x_valseq = np.load(r"../data/seq/x_val_seq.npy", allow_pickle=True)
```

```python
def create_sequences(length):
    x_train_pad = pad_sequences(
        x_trainseq,
        maxlen=length,
        padding="post",
        truncating="post"
    )

    x_val_pad = pad_sequences(
        x_valseq,
        maxlen=length,
        padding="post",
        truncating="post"
    )

    x_test_pad = pad_sequences(
        x_testseq,
        maxlen=length,
        padding="post",
        truncating="post"
    )
    return x_train_pad, x_val_pad, x_test_pad
```
